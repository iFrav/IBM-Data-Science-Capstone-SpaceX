# SpaceX Falcon 9 - EDA with SQL
Use SQL aggregation to answer targeted questions about launch sites, payloads and landing outcomes.

```python
import pandas as pd
import sqlite3
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv'
df = pd.read_csv(URL)
conn = sqlite3.connect(':memory:')
df.to_sql('SPACEXTBL', conn, index=False, if_exists='replace')
df.head()
```

```python
pd.read_sql_query('SELECT DISTINCT "Launch_Site" FROM SPACEXTBL', conn)
```

```python
q = '''SELECT SUM("PAYLOAD_MASS__KG_") AS total_payload_kg FROM SPACEXTBL WHERE "Customer" LIKE '%NASA (CRS)%' '''
pd.read_sql_query(q, conn)
# Course dataset result used in the presentation: 45,596 kg
```

```python
q = '''SELECT AVG("PAYLOAD_MASS__KG_") AS avg_payload_kg FROM SPACEXTBL WHERE "Booster_Version" = 'F9 v1.1' '''
pd.read_sql_query(q, conn)
# Course dataset result used in the presentation: 2,928.4 kg
```

```python
q = '''SELECT MIN("Date") AS first_successful_ground_landing FROM SPACEXTBL WHERE "Landing_Outcome" LIKE 'Success%Ground%' '''
pd.read_sql_query(q, conn)
# Course dataset result used in the presentation: 2015-12-22
```
