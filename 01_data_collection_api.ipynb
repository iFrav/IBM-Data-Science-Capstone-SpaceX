# SpaceX Falcon 9 - Data Collection with API
Collect launch records from the SpaceX REST API and flatten the nested response into a modeling table.

```python
import requests
import pandas as pd

api_url = 'https://api.spacexdata.com/v4/launches/past'
launches = requests.get(api_url, timeout=30).json()
len(launches)
```

```python
rows = []
for launch in launches:
    rows.append({
        'FlightNumber': launch.get('flight_number'),
        'Date': launch.get('date_utc'),
        'Rocket': launch.get('rocket'),
        'LaunchSite': launch.get('launchpad'),
        'Payloads': launch.get('payloads'),
        'Cores': launch.get('cores'),
        'Success': launch.get('success'),
    })
df = pd.DataFrame(rows)
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce').dt.date
df = df[df['Date'] <= pd.to_datetime('2020-11-13').date()].copy()
df.to_csv('spacex_api_raw.csv', index=False)
df.shape
```
