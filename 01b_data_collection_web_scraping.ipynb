# SpaceX Falcon 9 - Data Collection with Web Scraping
Extract historical Falcon 9 launch records from an HTML table as a second data collection method.

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

URL = 'https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches'
html = requests.get(URL, timeout=30).text
soup = BeautifulSoup(html, 'html.parser')
tables = pd.read_html(str(soup))
len(tables)
```

```python
launch_tables = []
for table in tables:
    cols = [str(c) for c in table.columns]
    if any('Flight' in c for c in cols) and any('Payload' in c for c in cols):
        launch_tables.append(table)
len(launch_tables)
```

```python
scraped = pd.concat(launch_tables, ignore_index=True, sort=False) if launch_tables else pd.DataFrame()
scraped.to_csv('spacex_web_scraped.csv', index=False)
scraped.head()
```
