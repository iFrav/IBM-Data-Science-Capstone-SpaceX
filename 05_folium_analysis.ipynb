# SpaceX Falcon 9 - Interactive Visual Analytics with Folium
Map launch sites and landing outcomes, then inspect the operational geography around each pad.

```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster, MousePosition
from folium.features import DivIcon
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv'
df = pd.read_csv(URL)
df.head()
```

```python
site_df = df[['Launch Site','Lat','Long']].drop_duplicates()
m = folium.Map(location=[29.56, -95.08], zoom_start=4)
for _, row in site_df.iterrows():
    folium.Circle(location=[row['Lat'], row['Long']], radius=1000, color='#d35400', fill=True, popup=row['Launch Site']).add_to(m)
    folium.Marker([row['Lat'], row['Long']], icon=DivIcon(icon_size=(120,24), html=f"<b>{row['Launch Site']}</b>")).add_to(m)
m
```

```python
cluster = MarkerCluster().add_to(m)
for _, row in df.iterrows():
    color = 'green' if int(row['class']) == 1 else 'red'
    folium.Marker([row['Lat'], row['Long']], icon=folium.Icon(color=color), popup=f"{row['Launch Site']} | class={row['class']}").add_to(cluster)
MousePosition().add_to(m)
m
```

## Interpretation
Florida contains closely spaced launch facilities, while VAFB is geographically separated on the US west coast. The map also makes it straightforward to inspect proximity to coastlines and transportation infrastructure, which matters for launch safety, logistics and recovery planning.
