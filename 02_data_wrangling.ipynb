# SpaceX Falcon 9 - Data Wrangling
Prepare the launch data for exploratory analysis and machine learning.

```python
import pandas as pd
DATA_URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv'
df = pd.read_csv(DATA_URL)
df.head()
```

```python
df['PayloadMass'] = df['PayloadMass'].fillna(df['PayloadMass'].mean())
df['LandingPad'] = df['LandingPad'].fillna('Unknown')
df.isna().sum()
```

```python
bad_outcomes = {'False Ocean','False RTLS','False ASDS','None ASDS','None None'}
df['Class'] = (~df['Outcome'].isin(bad_outcomes)).astype(int)
df[['Outcome','Class']].head(10)
```

```python
success_rate = df['Class'].mean()
print(f'Overall landing success rate: {success_rate:.1%}')
df.to_csv('dataset_part_2.csv', index=False)
```
