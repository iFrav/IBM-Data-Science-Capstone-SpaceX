# SpaceX Falcon 9 - EDA with Visualization
Explore how payload, flight number, orbit and launch site relate to first-stage landing success.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
DATA_URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv'
df = pd.read_csv(DATA_URL)
df.head()
```

```python
sns.catplot(y='PayloadMass', x='FlightNumber', hue='Class', data=df, aspect=3)
plt.xlabel('Flight Number')
plt.ylabel('Payload Mass (kg)')
plt.show()
```

```python
site_success = df.groupby('LaunchSite')['Class'].mean().sort_values(ascending=False)
display(site_success.to_frame('SuccessRate'))
site_success.plot(kind='bar', ylabel='Success rate', title='Landing success by launch site')
plt.show()
```

```python
orbit_success = df.groupby('Orbit')['Class'].mean().sort_values(ascending=False)
display(orbit_success.to_frame('SuccessRate'))
orbit_success.plot(kind='bar', ylabel='Success rate', title='Landing success by orbit')
plt.show()
```

```python
features = df[['FlightNumber','PayloadMass','Orbit','LaunchSite','Flights','GridFins','Reused','Legs','Block','ReusedCount','Serial']].copy()
features = pd.get_dummies(features, columns=['Orbit','LaunchSite','Serial'])
features.to_csv('dataset_part_3.csv', index=False)
features.shape
```
